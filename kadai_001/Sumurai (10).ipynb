```python

from openpyxl import Workbook
from datetime import datetime

# 新しいExcelブックを作成
wb = Workbook()

# アクティブなシートを取得
ws = wb.active

# セルに値を入力
ws["B2"] = "請求書"
ws["B4"] = "株式会社ABC"
ws["B5"] = "〒101-0022 東京都千代田区神田練塀町300"
ws["B6"] = "TEL:03-1234-5678 FAX:03-1234-5678"
ws["B7"] = "担当者名：鈴木一郎 様"
ws["F4"] = "NO."
ws["F5"] = "日付"
ws["G4"] = "0001"
ws["B15"] = "合計"
ws["B16"] = "消費税"
ws["B17"] = "税込合計"

# A1に現在の日付を入力
ws["G5"] = datetime.today()

# 表示形式（例：2026/04/17）
ws["G5"].number_format = "yyyy/mm/dd"

# 列A-Gの幅を11に設定
for col in ["A", "B", "C", "D", "E", "F", "G"]:
    ws.column_dimensions[col].width = 11

# ヘッダー
headers = ["商品名", "数量", "単価", "金額"]

# ヘッダー入力（B10:E10）
for col, value in enumerate(headers, start=2):  # B列=2
    ws.cell(row=10, column=col, value=value)

# データ（商品名・数量・単価）
data = [
    ["商品A", 2, 10000],
    ["商品B", 1, 15000]
]

# データ入力（11行目以降）
for i, row_data in enumerate(data, start=11):
    ws.cell(row=i, column=2, value=row_data[0])  # 商品名（B列）
    ws.cell(row=i, column=3, value=row_data[1])  # 数量（C列）
    ws.cell(row=i, column=4, value=row_data[2])  # 単価（D列）

    # 金額（E列）に数式
    ws.cell(row=i, column=5, value=f"=C{i}*D{i}")

ws["E11"] = "=C11*D11"
ws["E12"] = "=C12*D12"
ws["E13"] = "=E11+E12"
ws["E15"] = "=E13"
ws["E16"] = "=E15*0.1"
ws["E17"] = "=E15*1.1"


# ファイルを保存
wb.save("sample.xlsx")
```
