```python
import pandas as pd

# データ作成
df = pd.DataFrame({
    '日付': ['2023-05-17', '2023-05-18', '2023-05-19', '2023-05-20', '2023-05-21'],
    '社員名': ['山田', '佐藤', '鈴木', '田中', '高橋'],
    '売上': [100, 200, 150, 300, 250],
    '部門': ['メーカー', '代理店', 'メーカー', '商社', '代理店']
})

# 平均売上を計算
avg_sales = df['売上'].mean()

# 平均売上列
df['平均売上'] = avg_sales

# 業績ランク関数
def performance(sales):
    if sales >= avg_sales + 50:
        return 'A'
    elif sales >= avg_sales:
        return 'B'
    else:
        return 'C'

# 業績ランク列
df['業績ランク'] = df['売上'].apply(performance)

# Excel出力（openpyxl使用）
writer = pd.ExcelWriter('業績.xlsx', engine='openpyxl')
df.to_excel(writer, sheet_name='Sheet1', index=False)

# 列幅を11に設定
worksheet = writer.sheets['Sheet1']

for col in worksheet.columns:
    col_letter = col[0].column_letter  # 列名取得（A, B, C...）
    worksheet.column_dimensions[col_letter].width = 11

# 保存
writer.close()
```
